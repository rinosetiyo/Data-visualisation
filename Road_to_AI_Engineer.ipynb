Proyek mini tahap 1 : Analisis Nilai Siswa
dengan pembahasan tentang fondasi komputasi dan python
yang dipelajari adalah variabel, tipe data, daftar, operasi, dan logika

```python
nilai = [78, 65, 90, 82, 75]

jumlah_siswa = len(nilai)
print("Jumlah siswa:", jumlah_siswa)

jumlah = sum(nilai)
rata_rata = jumlah / len(nilai)
tertinggi = max(nilai)
terendah = min(nilai)

print("Jumlah nilai:", jumlah)
print("Rata-rata nilai:", rata_rata)
print("Nilai tertinggi:", tertinggi)
print("Nilai terendah:", terendah)

if rata_rata >= 80:
  score = "A"
  print("Siswa mendapatkan nilai", score)
elif rata_rata >= 70:
  score = "B"
  print("Siswa mendapatkan nilai", score)
elif rata_rata >= 60:
  score = "C"
  print("Siswa mendapatkan nilai", score)
elif rata_rata >= 50:
  score = "D"
  print("Siswa mendapatkan nilai", score)
```

Tahap 2
Matematika, Statistika, dan Analisis Data

```python
import statistics

pendapatan = [2,3,3,4,100]

print("Mean", statistics.mean(pendapatan))
print("Median", statistics.median(pendapatan))

# latihan mini mean dan median

waktu = [100, 105, 110, 115, 500]

print("Mean", statistics.mean(waktu))
print("Median", statistics.median(waktu))

# latihan mini range, varians (variance), simpang baku (standard deviation)

kelompok_a = [48, 49, 50, 51, 52]
kelompok_b = [10, 30, 50, 70, 90]

print("Range kelompok_a:", max(kelompok_a) - min(kelompok_a))
print("variance pupulasi a:", statistics.pvariance(kelompok_a))
print("simpang baku populasi a:", statistics.pstdev(kelompok_a))

print("Range kelompok_b:", max(kelompok_b) - min(kelompok_b))
print("variance pupulasi b:", statistics.pvariance(kelompok_b))
print("simpang baku populasi b:", statistics.pstdev(kelompok_b))

# probabilitas dasar

total_sesi = 200
sesi_gangguan = 12

probabilitas = sesi_gangguan / total_sesi

print("probabilitas empiris:", probabilitas)
print("persentase:", probabilitas * 100, "%")
```
